```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os

from pyunicorn.timeseries import RecurrencePlot

from pyunicorn.timeseries import RecurrenceNetwork

import logging
logging.basicConfig(level=logging.CRITICAL)
```

```python
%matplotlib inline
```

```python
FILE_NAME = r"D:\Pruebas%20BCI\SeñalesProcesadas10Hz2\AlejandoPayan\S3\10deTorquePre.csv"

selected_channels = [
    ["C1", "FCz", "Cz"],
    ["FC4", "C1", "Cz", "C1", "P3"],
    ["C3", "C2", "CP3"]
]
band = "mu"

if band == "mu":
    selected_channels = selected_channels[0]
elif band == "beta":
    selected_channels = selected_channels[1]
else:
    selected_channels = selected_channels[2]
```

```python
data = pd.read_csv(FILE_NAME)
binary_signal = data["Binaria"]
selected_columns = [f"{band}_" + ch for ch in selected_channels]
data = data[selected_columns] / data[selected_columns].max()

data.shape
```

```python
data.values
```

```python
rp = RecurrencePlot(
    data.values,
    threshold=data.std().mean() * 0.9,  # 0.12 * 2 como en tu código original
    #tau=3,
    epsilon='distance',
    #recurrence_rate = 0.09,
    metric="euclidean",
    dim=3  # Dimensión del espacio de fase
)
```

```python
plt.matshow(rp.recurrence_matrix())
plt.xlabel("$n$"); plt.ylabel("$n$")
```

```python
lorenz_rn = RecurrenceNetwork(data.values, threshold=data.std().mean() * 0.9, metric="supremum")
lorenz_C_v = lorenz_rn.local_clustering()
print(f"C_v = {lorenz_C_v}")
CD_v = np.log(lorenz_C_v) / np.log(3/4)
print(f"CD_v = {CD_v}")
```

```python
%matplotlib qt
```

```python
# plot trajectory on Lorenz attractor
ax = plt.figure().add_subplot(projection='3d')
ax.plot(*(data.values.T), lw=0.5, color="grey")
# plot local clustering dimension
scat_plot = ax.scatter(*(data.values.T), lw=0.1, c=CD_v, s=20, cmap=plt.colormaps["jet"])
cb = plt.colorbar(scat_plot, pad=0.2)
ax.set_xlabel("X"); ax.set_ylabel("Y"); ax.set_zlabel("Z")
ax.set_title(f"$CD_v$ for fixed $RR = {0.5}$");
```
